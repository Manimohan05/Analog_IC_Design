```python
import matplotlib.pyplot as plt
import numpy as np

# Data
divider_ratios = np.array([2, 3, 4, 5, 6, 8, 9, 10, 12, 15, 16])
frequencies = np.array([0.398969, 0.399927, 0.799182, 1.0015, 0.80176, 1.5844, 800.73, 1.99355, 1.60763, 2.01248, 3.1864])
locking_times = np.array([0.145, 0.16012, 0.185, 0.300, 0.32, 0.3401, 0.510, 0.535, 0.600, 0.625, 0.690])

# Masking the outlier in frequency (800.73) for a better fit
mask = frequencies < 10
filtered_ratios = divider_ratios[mask]
filtered_frequencies = frequencies[mask]

# Fit lines
freq_fit = np.polyfit(filtered_ratios, filtered_frequencies, 1)  # Linear fit
freq_line = np.poly1d(freq_fit)

lock_fit = np.polyfit(divider_ratios, locking_times, 1)  # Linear fit
lock_line = np.poly1d(lock_fit)

# Plotting
plt.figure(figsize=(12, 5))

# Plot 1: Divider Ratio vs Output Frequency
plt.subplot(1, 2, 1)
plt.scatter(divider_ratios, frequencies, label='Data', color='blue')
plt.plot(filtered_ratios, freq_line(filtered_ratios), label='Best Fit Line', color='red')
plt.xlabel('Divider Ratio (N)')
plt.ylabel('Output Frequency (GHz)')
plt.title('Divider Ratio vs Output Frequency')
plt.grid(True)
plt.legend()

# Plot 2: Divider Ratio vs Locking Time
plt.subplot(1, 2, 2)
plt.scatter(divider_ratios, locking_times, label='Data', color='green')
plt.plot(divider_ratios, lock_line(divider_ratios), label='Best Fit Line', color='orange')
plt.xlabel('Divider Ratio (N)')
plt.ylabel('Locking Time (µs)')
plt.title('Divider Ratio vs Locking Time')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
